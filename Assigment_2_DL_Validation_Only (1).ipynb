##Task 2: Defining the Network


**Network Architecture Definition**

```python
import torch.nn as nn
import torch.nn.functional as F

class CNN(nn.Module):
    def __init__(self):
        super(CNN, self).__init__()
        # Conv block 1: in_channels=3 (RGB), out_channels=32, kernel_size=3, stride=1, padding=1
        self.conv1 = nn.Conv2d(in_channels=3, out_channels=32, kernel_size=3, stride=1, padding=1)
        # Conv block 2: in_channels=32, out_channels=64, kernel_size=3, stride=1, padding=1
        self.conv2 = nn.Conv2d(in_channels=32, out_channels=64, kernel_size=3, stride=1, padding=1)

        # Maxpooling layer: kernel_size=2, stride=2
        self.pool = nn.MaxPool2d(kernel_size=2, stride=2)

        # Fully connected layers (Classifier)
        # Input image: 32x32 -> After pool 1: 16x16 -> After pool 2: 8x8 -> 64 channels * 8 * 8
        self.fc1 = nn.Linear(64 * 8 * 8, 128)
        self.fc2 = nn.Linear(128, 10)  # 10 output classes for CIFAR-10

    def forward(self, x):
        # Feature extraction using convolution, activation, and maxpooling
        x = self.pool(F.relu(self.conv1(x)))
        x = self.pool(F.relu(self.conv2(x)))

        # Flatten feature maps before feeding into the classifier
        x = x.view(x.size(0), -1)

        # Classification head
        x = F.relu(self.fc1(x))
        x = self.fc2(x)
        return x

# Instantiate the model
model = CNN()
print(model)
```

**Base Training & Testing Functions**

```python

import torch
import torch.nn as nn
import torch.optim as optim

# Training function
def train(model, trainloader, criterion, optimizer, device):
    model.train()
    running_loss = 0.0
    correct = 0
    total = 0

    for images, labels in trainloader:
        images = images.to(device)
        labels = labels.to(device)

        optimizer.zero_grad()
        outputs = model(images)
        loss = criterion(outputs, labels)

        loss.backward()
        optimizer.step()

        running_loss += loss.item() * images.size(0)
        _, predicted = torch.max(outputs, 1)
        total += labels.size(0)
        correct += (predicted == labels).sum().item()

    epoch_loss = running_loss / total
    epoch_accuracy = 100 * correct / total

    return epoch_loss, epoch_accuracy


# Testing function
def test(model, testloader, criterion, device):
    model.eval()
    running_loss = 0.0
    correct = 0
    total = 0

    with torch.no_grad():
        for images, labels in testloader:
            images = images.to(device)
            labels = labels.to(device)

            outputs = model(images)
            loss = criterion(outputs, labels)

            running_loss += loss.item() * images.size(0)
            _, predicted = torch.max(outputs, 1)
            total += labels.size(0)
            correct += (predicted == labels).sum().item()

    test_loss = running_loss / total
    test_accuracy = 100 * correct / total

    return test_loss, test_accuracy
```

## Task 4: Dataset Augmentation

Data augmentation is applied to the CIFAR-10 training dataset
to introduce variations in the images and improve the CNN model's
ability to generalize.

The techniques include horizontal flipping, random cropping,
color jittering, rotation, Gaussian noise, and normalization.
Image format conversions are also demonstrated.

```python
# Task 4: Dataset Augmentation

import torch
import torchvision
import torchvision.transforms as transforms
import matplotlib.pyplot as plt
from torch.utils.data import DataLoader

# Custom Gaussian Noise Transformation
class AddGaussianNoise:
    def __init__(self, mean=0.0, std=0.05):
        self.mean = mean
        self.std = std

    def __call__(self, tensor):
        noise = torch.randn_like(tensor) * self.std + self.mean
        return torch.clamp(tensor + noise, 0.0, 1.0)

print("Gaussian Noise transformation defined successfully!")
```

```python
# CIFAR-10 Normalization Parameters
CIFAR_MEAN = (0.4914, 0.4822, 0.4465)
CIFAR_STD = (0.2470, 0.2435, 0.2616)

# Training Transformations (standard, lighter CIFAR-10 augmentation)
train_transform = transforms.Compose([

    transforms.RandomCrop(32, padding=4),

    transforms.RandomHorizontalFlip(p=0.5),

    transforms.ToTensor(),

    transforms.Normalize(
        mean=CIFAR_MEAN,
        std=CIFAR_STD
    )
])

# Testing Transformations
test_transform = transforms.Compose([

    transforms.ToTensor(),

    transforms.Normalize(
        mean=CIFAR_MEAN,
        std=CIFAR_STD
    )
])

print("Data Augmentation transformations defined successfully!")
```

```python
# Load CIFAR-10 Training Dataset
augmented_trainset = torchvision.datasets.CIFAR10(
    root=DATA_PATH,
    train=True,
    download=False,
    transform=train_transform
)

# Load CIFAR-10 Testing Dataset
augmented_testset = torchvision.datasets.CIFAR10(
    root=DATA_PATH,
    train=False,
    download=False,
    transform=test_transform
)

# Create DataLoaders
augmented_trainloader = DataLoader(
    augmented_trainset,
    batch_size=32,
    shuffle=True
)

augmented_testloader = DataLoader(
    augmented_testset,
    batch_size=32,
    shuffle=False
)

# Display Dataset Information
print("Augmented training samples:", len(augmented_trainset))
print("Testing samples:", len(augmented_testset))

# Check Tensor Shapes
train_images, train_labels = next(iter(augmented_trainloader))
test_images, test_labels = next(iter(augmented_testloader))

print("Training batch shape:", train_images.shape)
print("Testing batch shape:", test_images.shape)
```

```python
# Visualize Different Data Augmentation Techniques

# Get original CIFAR-10 image
original_tensor, label = trainset[0]

# Convert Tensor to PIL Image
pil_image = transforms.ToPILImage()(original_tensor)

# Define Individual Augmentation Techniques
augmentation_methods = {

    "Original Image":
        transforms.Lambda(lambda img: img),

    "ToPILImage":
        transforms.ToPILImage(),

    "Horizontal Flip":
        transforms.RandomHorizontalFlip(p=1.0),

    "Random Resized Crop":
        transforms.RandomResizedCrop(
            size=32,
            scale=(0.8, 1.0)
        ),

    "Color Jitter":
        transforms.ColorJitter(
            brightness=0.4,
            contrast=0.4,
            saturation=0.4,
            hue=0.1
        ),

    "Random Rotation":
        transforms.RandomRotation(degrees=(15, 15)),

    "Gaussian Noise":
        AddGaussianNoise(mean=0.0, std=0.1),

    "Normalization":
        transforms.Normalize(
            mean=CIFAR_MEAN,
            std=CIFAR_STD
        ),

    "ToTensor":
        transforms.ToTensor()
}

# Create Figure
fig, axes = plt.subplots(3, 3, figsize=(12, 12))

for ax, (name, transform_fn) in zip(
    axes.flat,
    augmentation_methods.items()
):

    if name == "ToPILImage":
        transformed = transform_fn(original_tensor)
        display_image = transformed

    elif name == "Gaussian Noise":
        transformed = transform_fn(original_tensor)
        display_image = transformed.permute(1, 2, 0)

    elif name == "Normalization":
        transformed = transform_fn(original_tensor)

        # Reverse normalization for correct visualization
        mean = torch.tensor(CIFAR_MEAN).view(3, 1, 1)
        std = torch.tensor(CIFAR_STD).view(3, 1, 1)

        display_image = (
            transformed * std + mean
        ).clamp(0, 1).permute(1, 2, 0)

    elif name == "ToTensor":
        transformed = transform_fn(pil_image)
        display_image = transformed.permute(1, 2, 0)

    else:
        transformed = transform_fn(pil_image)
        display_image = transformed

    ax.imshow(display_image)
    ax.set_title(name)
    ax.axis("off")

plt.suptitle(
    "CIFAR-10 Data Augmentation Techniques",
    fontsize=16
)

plt.tight_layout()
plt.show()
```

##Task 5: Training Function and Tuning hyperparameters

```python
import torch
import torch.nn as nn
import torch.optim as optim
import time
import pandas as pd
import matplotlib.pyplot as plt

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

print("Using device:", device)

if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))

print("Training samples:", len(augmented_trainset))
print("Testing samples:", len(augmented_testset))

print("Training batches:", len(augmented_trainloader))
print("Testing batches:", len(augmented_testloader))
```

```python
# Deeper CNN Architecture
class CNN_Deep(nn.Module):
    def __init__(self):
        super(CNN_Deep, self).__init__()

        # Convolutional layers
        self.conv1 = nn.Conv2d(3, 32, kernel_size=3, padding=1)
        self.conv2 = nn.Conv2d(32, 64, kernel_size=3, padding=1)
        self.conv3 = nn.Conv2d(64, 128, kernel_size=3, padding=1)

        self.pool = nn.MaxPool2d(kernel_size=2, stride=2)

        # Fully connected layers
        self.fc1 = nn.Linear(128 * 4 * 4, 128)
        self.fc2 = nn.Linear(128, 10)

    def forward(self, x):

        x = self.pool(F.relu(self.conv1(x)))
        x = self.pool(F.relu(self.conv2(x)))
        x = self.pool(F.relu(self.conv3(x)))

        x = x.view(x.size(0), -1)

        x = F.relu(self.fc1(x))
        x = self.fc2(x)

        return x


# Improved CNN: VGG-style blocks + BatchNorm + Dropout
def conv_block(in_ch, out_ch):
    return nn.Sequential(
        nn.Conv2d(in_ch, out_ch, kernel_size=3, padding=1, bias=False),
        nn.BatchNorm2d(out_ch),
        nn.ReLU(inplace=True),
        nn.Conv2d(out_ch, out_ch, kernel_size=3, padding=1, bias=False),
        nn.BatchNorm2d(out_ch),
        nn.ReLU(inplace=True),
        nn.MaxPool2d(kernel_size=2, stride=2),
    )


class CNN_BN(nn.Module):
    def __init__(self):
        super(CNN_BN, self).__init__()

        # 32x32 -> 16x16 -> 8x8 -> 4x4
        self.features = nn.Sequential(
            conv_block(3, 64),
            conv_block(64, 128),
            conv_block(128, 256),
        )

        self.classifier = nn.Sequential(
            nn.Flatten(),
            nn.Dropout(0.4),
            nn.Linear(256 * 4 * 4, 256),
            nn.ReLU(inplace=True),
            nn.Dropout(0.4),
            nn.Linear(256, 10),
        )

    def forward(self, x):
        x = self.features(x)
        return self.classifier(x)


print("Deeper CNN architecture defined successfully!")
print(CNN_Deep())
print(CNN_BN())
```

```python
# Hyperparameter and Architecture Experiments

experiments = [
    {
        "name": "Exp 1 - Baseline",
        "model": CNN,
        "batch_size": 32,
        "lr": 0.001
    },
    {
        "name": "Exp 2 - Batch Size 64",
        "model": CNN,
        "batch_size": 64,
        "lr": 0.001
    },
    {
        "name": "Exp 3 - Learning Rate 0.0005",
        "model": CNN,
        "batch_size": 32,
        "lr": 0.0005
    },
    {
        "name": "Exp 4 - Deeper CNN",
        "model": CNN_Deep,
        "batch_size": 32,
        "lr": 0.001
    },
    {
        "name": "Exp 5 - Deeper CNN Tuned",
        "model": CNN_Deep,
        "batch_size": 64,
        "lr": 0.0005
    }
]

results = []

EPOCHS = 3

for exp in experiments:

    print("\n" + "=" * 50)
    print(exp["name"])
    print("=" * 50)

    trainloader = DataLoader(
        augmented_trainset,
        batch_size=exp["batch_size"],
        shuffle=True,
        pin_memory=True
    )

    testloader = DataLoader(
        augmented_testset,
        batch_size=exp["batch_size"],
        shuffle=False,
        pin_memory=True
    )

    model = exp["model"]().to(device)

    criterion = nn.CrossEntropyLoss()

    optimizer = optim.Adam(
        model.parameters(),
        lr=exp["lr"]
    )

    start_time = time.time()

    for epoch in range(EPOCHS):

        train_loss, train_acc = train(
            model, trainloader, criterion, optimizer, device
        )

        test_loss, test_acc = test(
            model, testloader, criterion, device
        )

        print(
            f"Epoch [{epoch+1}/{EPOCHS}] "
            f"Train Loss: {train_loss:.4f} | "
            f"Train Acc: {train_acc:.2f}% | "
            f"Test Loss: {test_loss:.4f} | "
            f"Test Acc: {test_acc:.2f}%"
        )

    elapsed_time = time.time() - start_time

    results.append({
        "Experiment": exp["name"],
        "Architecture": exp["model"].__name__,
        "Batch Size": exp["batch_size"],
        "Learning Rate": exp["lr"],
        "Train Loss": train_loss,
        "Train Accuracy": train_acc,
        "Test Loss": test_loss,
        "Test Accuracy": test_acc,
        "Training Time (seconds)": round(elapsed_time, 2)
    })

    print(f"Training Time: {elapsed_time:.2f} seconds")

    del model
    if torch.cuda.is_available():
        torch.cuda.empty_cache()

print("\nAll experiments completed!")
```

```python
# Experiment 7: Improved CNN (BatchNorm + Dropout, light augmentation,
# AdamW + OneCycleLR, label smoothing, 30 epochs)

EPOCHS_7 = 30
BATCH_7 = 128

# ONLY NEW CHANGE: held-out validation subset of the official 50k training images.
# The CNN, optimizer, scheduler, augmentation and other experiments are unchanged.
import numpy as np
from pathlib import Path
from sklearn.model_selection import train_test_split
from torch.utils.data import Subset

split_indices7 = np.arange(len(augmented_trainset))
train_indices7, val_indices7 = train_test_split(
    split_indices7,
    test_size=0.10,
    stratify=np.asarray(augmented_trainset.targets),
    random_state=42
)

# Validation uses test_transform, so it has NO random training augmentation.
validation_source7 = torchvision.datasets.CIFAR10(
    root=DATA_PATH, train=True, download=False, transform=test_transform
)
training_subset7 = Subset(augmented_trainset, train_indices7.tolist())
validation_subset7 = Subset(validation_source7, val_indices7.tolist())

assert len(training_subset7) == 45000
assert len(validation_subset7) == 5000
assert len(augmented_testset) == 10000
assert set(train_indices7).isdisjoint(set(val_indices7))
assert np.array_equal(augmented_trainset.targets, validation_source7.targets)
print('Exp 7 split | Train: 45,000 | Validation: 5,000 | Test: 10,000')
print('Train/Validation overlap:', len(set(train_indices7) & set(val_indices7)))

best_checkpoint_path7 = Path('exp7_best_validation_model.pt')
best_val_accuracy7 = -1.0
best_val_loss7 = float('inf')
best_epoch7 = None

train_loader7 = DataLoader(
    training_subset7, batch_size=BATCH_7, shuffle=True,
    num_workers=2, pin_memory=True
)
val_loader7 = DataLoader(
    validation_subset7, batch_size=BATCH_7, shuffle=False,
    num_workers=2, pin_memory=True
)
test_loader7 = DataLoader(
    augmented_testset, batch_size=BATCH_7, shuffle=False,
    num_workers=2, pin_memory=True
)

model_exp7 = CNN_BN().to(device)

criterion_train = nn.CrossEntropyLoss(label_smoothing=0.1)
criterion_eval = nn.CrossEntropyLoss()   # plain loss, comparable to earlier experiments

optimizer = optim.AdamW(model_exp7.parameters(), lr=2e-3, weight_decay=5e-4)
scheduler = optim.lr_scheduler.OneCycleLR(
    optimizer,
    max_lr=2e-3,
    epochs=EPOCHS_7,
    steps_per_epoch=len(train_loader7)
)

exp7_history = {
    "train_loss": [],
    "train_accuracy": [],
    "val_loss": [],
    "val_accuracy": [],
    "test_loss": [],
    "test_accuracy": []
}

start_time = time.time()

print("Experiment 7: CNN_BN - 30 Epochs")
print("=" * 60)

for epoch in range(EPOCHS_7):

    model_exp7.train()
    running_loss, correct, total = 0.0, 0, 0

    for images, labels in train_loader7:
        images, labels = images.to(device), labels.to(device)

        optimizer.zero_grad()
        outputs = model_exp7(images)
        loss = criterion_train(outputs, labels)
        loss.backward()
        optimizer.step()
        scheduler.step()          # step the scheduler once per batch

        running_loss += loss.item() * images.size(0)
        correct += (outputs.argmax(1) == labels).sum().item()
        total += labels.size(0)

    train_loss = running_loss / total
    train_acc = 100 * correct / total
    # Validation determines the selected checkpoint; never select by Test accuracy.
    val_loss, val_acc = test(model_exp7, val_loader7, criterion_eval, device)
    # Test metrics are logged as before for the doctor's Task 5 curves ONLY.
    test_loss, test_acc = test(model_exp7, test_loader7, criterion_eval, device)

    if (val_acc > best_val_accuracy7 or
            (val_acc == best_val_accuracy7 and val_loss < best_val_loss7)):
        best_val_accuracy7 = val_acc
        best_val_loss7 = val_loss
        best_epoch7 = epoch + 1
        torch.save(model_exp7.state_dict(), best_checkpoint_path7)

    exp7_history["train_loss"].append(train_loss)
    exp7_history["train_accuracy"].append(train_acc)
    exp7_history["val_loss"].append(val_loss)
    exp7_history["val_accuracy"].append(val_acc)
    exp7_history["test_loss"].append(test_loss)
    exp7_history["test_accuracy"].append(test_acc)

    print(
        f"Epoch [{epoch+1}/{EPOCHS_7}] | "
        f"Train Loss: {train_loss:.4f} | Train Acc: {train_acc:.2f}% | "
        f"Val Loss: {val_loss:.4f} | Val Acc: {val_acc:.2f}% | "
        f"Test Loss: {test_loss:.4f} | Test Acc: {test_acc:.2f}%"
    )

exp7_time = time.time() - start_time

print("=" * 60)
print(f"Total Training Time: {exp7_time:.2f} seconds")
# Final metrics use the BEST VALIDATION checkpoint, not the last or best-Test epoch.
assert best_checkpoint_path7.exists()
try:
    selected_weights7 = torch.load(best_checkpoint_path7, map_location=device, weights_only=True)
except TypeError:  # compatibility with older PyTorch versions
    selected_weights7 = torch.load(best_checkpoint_path7, map_location=device)
model_exp7.load_state_dict(selected_weights7)
final_test_loss7, final_test_acc7 = test(model_exp7, test_loader7, criterion_eval, device)
final_train_loss7 = exp7_history['train_loss'][best_epoch7 - 1]
assert abs(final_test_acc7 - exp7_history['test_accuracy'][best_epoch7 - 1]) < 1e-6
assert abs(final_test_loss7 - exp7_history['test_loss'][best_epoch7 - 1]) < 1e-4
print(f"Best Validation Accuracy: {best_val_accuracy7:.2f}% at Epoch {best_epoch7}")
print(f"Final Train Loss (selected epoch; label-smoothed): {final_train_loss7:.4f}")
print(f"Final Test Loss (validation-selected model): {final_test_loss7:.4f}")
print(f"Final Test Accuracy (validation-selected model): {final_test_acc7:.2f}%")
print(f"Checkpoint saved to: {best_checkpoint_path7}")
```
